# STUDENT PERFORMANCE ANALYZER
**Beginner-level B.Tech 1st Year Mini Project**

This project uses **Python**, **NumPy**, and **Pandas** to analyze student academic performance.

---

### Objectives
* Store student marks in a Pandas DataFrame
* Calculate total and average marks
* Assign grades
* Find the highest and lowest performers
* Calculate subject-wise averages
* Identify students who need improvement
* Create simple visualizations

**Libraries used:** Python, NumPy, Pandas, Matplotlib

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

### 1. Create the Student Dataset
We create a small sample dataset containing student names and marks in five subjects.

You can replace the sample values with your own class data later.

```python
# Sample student data
data = {
    "Name": ["Aarav", "Priya", "Rahul", "Sneha", "Arjun",
             "Ananya", "Rohan", "Neha", "Vikash", "Kavya"],
    "Python": [85, 72, 91, 68, 78, 95, 60, 88, 74, 82],
    "Mathematics": [78, 81, 89, 65, 72, 92, 58, 84, 70, 79],
    "Physics": [82, 76, 94, 70, 75, 90, 62, 86, 73, 81],
    "English": [88, 79, 85, 74, 80, 93, 67, 90, 76, 84],
    "Programming": [90, 85, 96, 69, 82, 98, 64, 92, 78, 87]
}

df = pd.DataFrame(data)

print("Student Dataset:")
display(df)
```

### 2. Basic Information About the Dataset

```python
# Display basic information
print("Number of students:", len(df))
print("Number of subjects:", len(df.columns) - 1)
print("\nColumn names:")
print(df.columns.tolist())
print("\nDataset information:")
df.info()
```

### 3. Calculate Total and Average Marks

```python
# Calculate total and average marks
subjects = ["Python", "Mathematics", "Physics", "English", "Programming"]

df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df[subjects].mean(axis=1)

display(df[["Name"] + subjects + ["Total", "Average"]])
```

### 4. Assign Grades

```python
# Function to assign grade
def get_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"

df["Grade"] = df["Average"].apply(get_grade)

display(df[["Name", "Average", "Grade"]])
```

### 5. Find the Top and Lowest Performers

```python
# Student with highest average
top_student = df.loc[df["Average"].idxmax()]

# Student with lowest average
lowest_student = df.loc[df["Average"].idxmin()]

print("Top Performer:")
print(top_student[["Name", "Average", "Grade"]])

print("\nLowest Performer:")
print(lowest_student[["Name", "Average", "Grade"]])
```

### 6. Subject-wise Average Using NumPy and Pandas

```python
# Calculate subject-wise averages
subject_averages = df[subjects].mean()
print("Subject-wise Average Marks:")
print(subject_averages)

# NumPy example
marks_array = df[subjects].to_numpy()
print("\nOverall average calculated using NumPy:", np.mean(marks_array))
```

### 7. Students Who Need Improvement

```python
# Students with average below 60
needs_improvement = df[df["Average"] < 60]

if needs_improvement.empty:
    print("No student has an average below 60.")
else:
    display(needs_improvement[["Name", "Average", "Grade"]])
```

### 8. Sort Students by Performance

```python
# Sort students from highest to lowest average
ranking = df.sort_values(by="Average", ascending=False)
print("Student Performance Ranking:")
display(ranking[["Name", "Total", "Average", "Grade"]])
```

### 9. Visualize Subject-wise Average Marks

```python
# Bar chart of subject-wise average marks
subject_averages.plot(kind="bar", figsize=(8, 5))
plt.title("Subject-wise Average Marks")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 10. Visualize Student Average Performance

```python
# Bar chart of each student's average
plt.figure(figsize=(10, 5))
plt.bar(df["Name"], df["Average"])
plt.title("Student Average Performance")
plt.xlabel("Students")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.ylim(0, 100)
plt.tight_layout()
plt.show()
```

### 11. NumPy Statistical Analysis

```python
# NumPy statistics on total marks
total_marks = df["Total"].to_numpy()

print("Mean total marks:", np.mean(total_marks))
print("Maximum total marks:", np.max(total_marks))
print("Minimum total marks:", np.min(total_marks))
print("Standard deviation:", np.std(total_marks))
```

### 12. Save the Analyzed Dataset

```python
# Save the final analyzed data as a CSV file
df.to_csv("student_performance_analysis.csv", index=False)
print("File saved as: student_performance_analysis.csv")
```

The program calculates totals, averages, grades, rankings, subject-wise performance, and basic statistics. It also creates simple charts to make the results easier to understand.
